```python
nm_ids = [
    398400810,
    398400809,
    159221368,
    170598570,
    159222686,
    170602860,
    242981834,
    222683424,
    158994099,
    224050073,
    224050074
]
data = await fetch_get_price(nm_ids)
```

```python
# 1. Создаем пустой список для хранения всех строк
all_rows = []

for item in data:
    # 2. Извлекаем данные
    nm_id = item.get('nmID')
    vendor_code = item.get('vendorCode')
    
    # достаем цену из первого размера
    sizes = item.get('sizes', [])
    if sizes:
        price = sizes[0].get('price')
        discounted_price = sizes[0].get('discountedPrice')
    else:
        price = None
        discounted_price = None

    # 3. Создаем словарь для каждой итерации
    row = {
        'nm_id': nm_id,
        'vendor_code': vendor_code,
        'price': price,
        'discounted_price': discounted_price
    }
    
    # 4. Добавляем этот словарь в наш список
    all_rows.append(row)

# 5. Создаем DataFrame из списка словарей
# Убираем index=False, Pandas сам проставит индексы 0, 1, 2...
df = pd.DataFrame(all_rows)
```
